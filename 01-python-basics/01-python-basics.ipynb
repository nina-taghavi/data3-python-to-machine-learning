# Python Basics

## Variables and data types

```python
from keras.src.ops import dtype

# List of integers
numbers = [1,2,3,4,5]
print(numbers)

# List of strings
fruits = ['apple', 'banana', 'cherry']
print(fruits)

# Mixed-type list
mixed_lest = [1, 'apple', 3.5, [2,3]]
print(mixed_lest)
print(type(mixed_lest))

# Nested list (list with a list)
nested_list = [[1,2], [3,4], [5,6]]
print(nested_list)
```

```python
# int
my_variable = 1
print(my_variable)

# str
my_variable = "Hello world"
print(my_variable)

# bool
my_variable = True
print(my_variable)

# float
my_variable = 18.275
print(my_variable)

# print formatted string
print(f'The temperature is {my_variable:.1f} degrees')

# type function
print(type(1.2))
print(type(3>2))
```

```python
# list
list1 = [0, 1.0, "two"]
print(list1[1])
list1[1] = 2.0
print(list1[1])
```

```python
# tuple
tuple1 = (0, 1.0, "two")
print(tuple1[1])
tuple1[1] = 2.0  # error, tuple is immutable (unchangeable)
```

```python
# set
set1 = {5, "six"}
print(set1[1]) # error, set has no order£
```

```python
# dict
dict1 = {"f":1.0, "s":"six"}
print(dict1["s"])

# 2-dimensional list
a = [[1, 2, 3],
     [4, 5, 6],
     [7, 8, 9]]
print(a)
print(a[0])
a[0][1] = 10 # 2 replaced by 10, first index = row, second = column
print(a[0]) 	# [1, 10, 3]
```

## List elements and slicing

```python
list = [1, 2, 3, 4, 5, 6, 7 ]
```

```python
list 	       # [1, 2, 3, 4, 5, 6, 7]
```

```python
list[0]        # 1 (first element)
```

```python
list[-2]       # 6 (last but one element)
```

```python
list[1:4]      # [2, 3, 4] (from element 2 to element 4)
```

```python
list[:3]       # [1, 2, 3] (first 3 elements)
```

```python
list[-2:]      # [6, 7] (last 2 elements)
```

```python
list[:]        # [1, 2, 3, 4, 5, 6, 7] (all elements)
```

```python
list[3::2]     # [4, 6] (from element 4, step = 2)
```

```python
list[::-1]     # [7, 6, 5, 4, 3, 2, 1] (from last element, step -1)
```

## List comprehension

```python
[i for i in range(5)]
# [0, 1, 2, 3, 4]
```

```python
[x**2 for x in range(1,3)]
# [1, 4]
```

```python
[[x,x**2] for x in range(0,3,2)]
# [[0, 0], [2, 4]] (step 2)
```

```python
[w[0] for w in ["the", "world"]]
# ['t', 'w']
```

```python
[x+y for x in [10,30,50] for y in [2,4]]
#[12, 14, 32, 34, 52, 54]
```

## Program flow

```python
for i in range(4,12,2): # 4 -> 11, step 2
     print(i**2)
```

```python
if i <0:
     print("negative")
elif i == 0:
     print("zero")
else:
     print("positive")
```

## Functions

```python
def fac(n):
     if n <= 1:
          return 1
     else:
          return n * fac(n-1)
```

```python
fac(3)
```

```python
def power(base, exponent=2): # optional parameters last
     return base ** exponent
```

```python
power(2)
```

```python
power(5,1)
```

```python
power(base=5, exponent=1)
```

```python
power(exponent=1, base=5)
```
