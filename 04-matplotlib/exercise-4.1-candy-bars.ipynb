Create a scatterplot for 'sugarpercent' and 'winpercent'. Create a figure for the scatter plot with a size of 10 by 6 inches.
Plot a scatter plot using the 'sugarpercent' column for the x-axis and the 'winpercent' column for the y-axis. Use the color 'skyblue' for the points, set the size of the markers to 200, and apply a black edge color.

Add a title to the scatter plot that reads "Sugar Percentage vs Winning Percentage of Candy Bars". Label the x-axis as "Sugar Percentage (%)" and the y-axis as "Winning Percentage (%)".

Annotate the data points with the name of the candies.

Enable the grid in the scatter plot for better readability of data points.

Finally, display the plot using the appropriate function.

```python
# Create a scatter plot
plt.figure(figsize=(10, 6))  # Set the figure size
plt.scatter(df['sugarpercent'], df['winpercent'], color='skyblue', s=200, edgecolor='black')

# Add titles and labels
plt.title('Sugar Percentage vs Winning Percentage of Candy Bars')
plt.xlabel('Sugar Percentage (%)')
plt.ylabel('Winning Percentage (%)')

# Annotate points with competitor names (Additional step - for your information)
for i in range(len(df)):
    plt.annotate(
        df['competitorname'][i],  # Text label
        (df['sugarpercent'][i], df['winpercent'][i]),  # Coordinates (x, y)
        textcoords="offset points",  # Position relative to point
        xytext=(0, 10),  # Offset: 10 points above
        ha='center'  # Horizontal alignment
    )

# Show grid and plot
plt.grid(True)
plt.show()
```

Create a scatterplot for 'pricepercent' and 'winpercent'. Create a figure for the scatter plot with a size of 10 by 6 inches.
Plot a scatter plot using the 'pricepercent' column for the x-axis and the 'winpercent' column for the y-axis. Set the size of the markers to 200, and apply a black edge color.

Color the datapoints blue for the candy bars that contain chocolate, and orange for the candy bars that do not contain chocolate).

Add a title to the scatter plot that reads "Price Percentage vs Winning Percentage of Candy Bars". Label the x-axis as "Price Percentage (%)" and the y-axis as "Winning Percentage (%)".

Annotate the data points with the name of the candies.

Enable the grid in the scatter plot for better readability of data points.

Finally, display the plot using the appropriate function.

```python
# Create a scatter plot
plt.figure(figsize=(10, 6))  # Set the figure size

# Color-coding based on the 'chocolate' column
colors = np.where(df['chocolate'] == 'Yes', 'blue', 'orange')
# print(colors)
plt.scatter(df['pricepercent'], df['winpercent'], color=colors, s=200, edgecolor='black')

# Add titles and labels
plt.title('Price Percentage vs Winning Percentage of Candy Bars')
plt.xlabel('Price Percentage (%)')
plt.ylabel('Winning Percentage (%)')

# Annotate points with competitor names
for i in range(len(df)):
    plt.annotate(df['competitorname'][i], (df['pricepercent'][i], df['winpercent'][i]),
                 textcoords="offset points", xytext=(0, 10), ha='center')

# Show grid and plot
plt.grid(True)
plt.show()
```

```python
# Create a scatter plot
plt.figure(figsize=(10, 6))  # Set the figure size

# Color-coding based on 'chocolate' column
colors = np.where(df['chocolate'] == 'Yes', 1, 0)  # Numeric values for colormap

plt.scatter(df['pricepercent'], df['winpercent'], c=colors, s=200, edgecolor='black')

# Add titles and labels
plt.title('Price Percentage vs Winning Percentage of Candy Bars')
plt.xlabel('Price Percentage (%)')
plt.ylabel('Winning Percentage (%)')

# Annotate points with competitor names
for i in range(len(df)):
    plt.annotate(df['competitorname'][i], (df['pricepercent'][i], df['winpercent'][i]),
                 textcoords="offset points", xytext=(0, 10), ha='center')

# Show grid and plot
plt.grid(True)

plt.colorbar(label='Chocolate (1=Yes, 0=No)')
plt.tight_layout()
plt.show()
```
