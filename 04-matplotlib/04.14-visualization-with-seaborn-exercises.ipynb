b) Using Seaborn and Matplotlib, create a bar chart based on randomly generated data.

- Create a Python list with 5 categories labeled 'A', 'B', 'C', 'D', 'E'.
- Generate 5 random integer values between 1 and 20 by creating an rng using NumPy and using rng.integers().
- Use Seaborn's barplot to create a bar chart.
- Add a title and axis labels using Matplotlib.

```python
#SOLUTION_START

import numpy as np

# Generate random data
rng = np.random.default_rng(42)
categories = ['A', 'B', 'C', 'D', 'E']
values = rng.integers(1, 20, size=5)

# Create a bar plot
sns.barplot(x=categories, y=values)

# Adding a title and labels
plt.title("Random Bar Chart with Seaborn")
plt.xlabel("Category")
plt.ylabel("Values")

# Display the plot
plt.show()

#SOLUTION_END
```

### 2. Barchart with cities dataset

Import needed libraries and load cities dataset.

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

# Load dataset
cities  = pd.read_csv('data/california_cities.csv', sep=',', index_col=0)
cities.head()
```

Create a barchart to display the population per city. 
    
- Only display the 5 cities with the highest population.
- Make the bars a purple color

```python
#SOLUTION_START
# Sort the dataset by population in descending order and select the top 5 cities
top_cities = cities.sort_values(by='population_total', ascending=False).head(5)

#Create the barchart
sns.barplot(top_cities, x='city', y='population_total', color='purple')
#or alternatively use can pass the columns directly to x and y
#sns.barplot(x=top_cities['city'], y=top_cities['population_total'], color='purple')
plt.xlabel("city")
plt.ylabel("population")
plt.title("population per city")
plt.show()

#SOLUTION_END
```

### 3. Simple Boxplot with Seaborn
Create a boxplot to display some more randomly generated data
- Generate 20 random integer values between 1 and 100 using the method from the first exercise
- Use Seaborn's boxplot to create a boxplot.



```python
#SOLUTION_START
# Generate random data
rng = np.random.default_rng(42)
data = rng.integers(1, 100, size=20)

# Create a boxplot
sns.boxplot(y=data)
# Or if you want to flip the boxplot set x=data instead of y

# Adding a title and labels
plt.title("Random Boxplot with Seaborn")

# Display the plot
plt.show()
#SOLUTION_END
```
