# 📊 Regression

## 1. What Is Regression?

Regression is a supervised learning technique that predicts a **numerical outcome** from input features.

- **Goal:** fit a mathematical function that models the relationship between the features and the target.
- **Types:**
  - **Linear regression** — fits a straight line (or a flat plane/hyperplane with several features).
  - **Polynomial regression** — fits a curve (quadratic, cubic, …).

---

## 2. Linear Regression

### One predictor
$$y = ax + b$$
- $a$: slope — change in $y$ when $x$ increases by 1
- $b$: intercept — value of $y$ when $x = 0$

### Multiple predictors
$$y = b + a_1x_1 + a_2x_2 + \dots + a_nx_n$$
- $x_1, \dots, x_n$: features (predictors)
- $a_1, \dots, a_n$: coefficients — the effect of each feature
- $b$: intercept (bias term)

📌 Each coefficient $a_i$ is the change in $y$ when $x_i$ increases by one unit **while all other features stay constant**. Coefficients are only comparable with each other if the features are on the same scale.

---

## 3. `LinearRegression` in scikit-learn

| Hyperparameter | Default | Meaning |
| --- | --- | --- |
| `fit_intercept` | `True` | Whether to estimate the intercept $b$ |
| `copy_X` | `True` | Copy `X` instead of possibly overwriting it |
| `n_jobs` | `None` | Number of CPU cores to use |
| `positive` | `False` | Force all coefficients to be ≥ 0 |

After fitting: `model.intercept_` is $b$ and `model.coef_` holds $a_1, \dots, a_n$.

---

## 4. Workflow

1. **Prepare data:** predictors in a DataFrame `X` (2-D), target in a Series `y`.
2. **Split** into training and test sets (e.g. 80/20).
3. **Train:** `model.fit(X_train, y_train)`, then read the intercept and slope(s).
4. **Validate** on the test set with **R²** (`model.score`) plus MAE, MAPE or RMSE (formulas in `05.03`).
5. **Predict** for new data: `model.predict(X_new)`.
6. **Visualise:** a scatter plot with the fitted line shows whether the relationship is really linear; a residual plot (errors vs predictions) should show no pattern.

## 5. Polynomial Regression

$$y = a_0 + a_1x + a_2x^2 + \dots + a_nx^n$$

- **Quadratic:** $n = 2$  **Cubic:** $n = 3$
- Captures **non-linear** relationships while still being a *linear* model in its coefficients — so `LinearRegression` can fit it once the powers of $x$ are added as features.
- Higher degrees fit the training data more closely but overfit quickly; choose the degree by validation score.

```python
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import PolynomialFeatures

poly_model = make_pipeline(PolynomialFeatures(degree=3), LinearRegression())
poly_model.fit(X_train, y_train)
print("R²:", poly_model.score(X_test, y_test))
```

---

## ✅ Summary
- Regression predicts numerical outcomes.
- Linear regression is simple and interpretable.
- Polynomial regression captures curved patterns but needs its degree tuned to avoid overfitting.
- Evaluate with R², MAE, MAPE and RMSE on held-out data.
