## 1. Levels of Measurement (NOIR)

Stanley Smith Stevens' four levels of measurement — each level adds a property to the one before it:

$$\textbf{Nominal} \longrightarrow \textbf{Ordinal} \longrightarrow \textbf{Interval} \longrightarrow \textbf{Ratio}$$

| Level | Data type | Key property | Allowed operations | Example |
| --- | --- | --- | --- | --- |
| **Nominal** | Categorical | Names / labels only; no order | Counts, frequencies, mode | Blood type, country, OS |
| **Ordinal** | Categorical | Meaningful order, but gaps not equal | + median, percentiles, ranking | Rating (Bad, OK, Good), education level |
| **Interval** | Numerical | Order + **equal intervals**, **no true zero** | + addition, subtraction, mean | Temperature in °C or °F, calendar year |
| **Ratio** | Numerical | Order + equal intervals + **true zero** | + multiplication, division (ratios) | Weight, salary, execution time (ms), distance |

### Rule of thumb
- **Nominal:** just names.
- **Ordinal:** names + order.
- **Interval:** order + measurable, equal distances — but $0^\circ\text{C}$ is not "no heat", so $20^\circ\text{C}$ is not "twice as warm" as $10^\circ\text{C}$.
- **Ratio:** a true zero exists — 0 seconds means no time, and 20 seconds is twice as long as 10.

## 2. Correlation vs Causation

**Correlation is an observation; causation is an explanation.**

- **Correlation:** two variables move together — when $X$ changes, $Y$ tends to change too.
- **Causation:** one variable directly produces a change in the other — changing $X$ **makes** $Y$ change ($X \rightarrow Y$).

> **Golden rule:** *correlation does not imply causation.* Two metrics sharing a pattern does not mean one drives the other.

### Why correlation ≠ causation
When $X$ and $Y$ are strongly correlated, four explanations are possible:

1. **Direct causation ($X \rightarrow Y$)**
   - *Example:* higher CPU load directly increases server power consumption.
2. **Reverse causation ($Y \rightarrow X$)**
   - You assume $X$ causes $Y$, but $Y$ actually causes $X$.
   - *Example:* traffic and latency are correlated. Latency doesn't attract traffic; heavy traffic slows the server.
3. **Confounding variable ($Z \rightarrow X$ and $Z \rightarrow Y$)**
   - A hidden third factor drives both.
   - *Example:* ice-cream sales and drownings correlate. Ice cream doesn't cause drowning; **hot weather ($Z$)** makes people both buy ice cream and go swimming.
4. **Spurious correlation (coincidence)**
   - Two unrelated trends happen to line up by chance.
   - *Example:* US spending on science and suicides by hanging correlated at about 0.99 over a decade.

### Key differences

| Feature | Correlation | Causation |
| --- | --- | --- |
| **Meaning** | Association / co-movement | Cause-and-effect link |
| **Direction** | Symmetric: $X$ correlates with $Y \iff Y$ correlates with $X$ | Asymmetric: $X$ causing $Y$ does not mean $Y$ causes $X$ |
| **Measured by** | Pearson's $r$, Spearman's $\rho$, mutual information | Randomised experiments (A/B tests, RCTs), causal inference methods |
| **Intervention** | Changing $X$ may not change $Y$ at all | Changing $X$ predictably changes $Y$ |
| **Useful for** | Prediction and forecasting | Decisions and interventions (policy, design changes) |

### Establishing causation
Three conditions must hold for $X$ to cause $Y$:

1. **Temporal precedence:** the cause ($X$) happens before the effect ($Y$).
2. **Covariation:** when $X$ changes, $Y$ changes consistently.
3. **Non-spuriousness:** plausible alternative explanations and confounders are ruled out.

Common methods:
- **Randomised controlled trials (A/B tests):** random assignment to control and treatment groups balances out hidden confounders ($Z$).
- **Natural experiments & instrumental variables:** use external shocks (a policy change, an outage) to isolate the effect when an experiment isn't possible.
- **Difference-in-differences (DiD):** compare the change over time in an affected group with the change in an unaffected control group.

## 3. Measures of Central Tendency

| Measure | Definition | Best used for | Minimum level of measurement |
| --- | --- | --- | --- |
| **Mode** | Most frequent value | Categorical or discrete data | Nominal |
| **Median** | Middle value when sorted | Ordinal data, or skewed numbers / data with outliers | Ordinal (needs ordering) |
| **Mean** | Arithmetic average | Symmetric numerical data without extreme outliers | Interval (needs addition and division) |

In a right-skewed distribution (e.g. incomes), typically mode < median < mean — the mean is pulled towards the long tail.

## 4. Association Rules: Support

In **association rule learning** (e.g. the Apriori algorithm, market basket analysis), **support** measures **how often an itemset appears in the dataset** — its baseline popularity across all transactions.

### Formula
For an itemset $A$, or a rule $A \rightarrow B$, with $N$ transactions:

$$\text{Support}(A) = \frac{\text{transactions containing } A}{N}$$

$$\text{Support}(A \rightarrow B) = \frac{\text{transactions containing both } A \text{ and } B}{N} = P(A \cap B)$$

- Range: **0 to 1** (or 0% to 100%).
- Notation: $P(A \cap B)$ means "a transaction contains both $A$ and $B$". Itemset notation writes the same thing as $\text{Support}(A \cup B)$ — the union of the **items**.

### Example
A supermarket has **5 transactions**:

1. `{Milk, Bread}`
2. `{Milk, Diapers, Beer, Bread}`
3. `{Milk, Diapers, Beer}`
4. `{Bread, Diapers, Beer}`
5. `{Milk, Bread, Diapers, Beer}`

- **`{Milk}`** appears in transactions 1, 2, 3, 5:
  $$\text{Support}(\text{Milk}) = \tfrac{4}{5} = 0.8 \;(80\%)$$
- **`{Milk, Beer}`** appears in transactions 2, 3, 5:
  $$\text{Support}(\text{Milk} \rightarrow \text{Beer}) = \tfrac{3}{5} = 0.6 \;(60\%)$$
- **`{Milk, Bread, Beer}`** appears in transactions 2, 5:
  $$\text{Support}(\{\text{Milk}, \text{Bread}\} \rightarrow \text{Beer}) = \tfrac{2}{5} = 0.4 \;(40\%)$$

### Why support matters
1. **Filtering noise (minimum support, `min_sup`)**
   - Rare items can form coincidental rules (e.g. one caviar purchase alongside a particular wine).
   - A minimum support threshold (e.g. `min_sup` = 5%) removes these rare cases up front.
2. **Pruning the search**
   - **Apriori principle:** if an itemset is infrequent (support below `min_sup`), all of its supersets are infrequent too — so they never need to be counted.

### The three core metrics

| Metric | Formula | Measures |
| --- | --- | --- |
| **Support** | $P(A \cap B)$ | **Frequency:** how common the combination is overall |
| **Confidence** | $\dfrac{P(A \cap B)}{P(A)}$ | **Reliability:** given $A$ was bought, how likely is $B$? |
| **Lift** | $\dfrac{\text{Confidence}(A \rightarrow B)}{P(B)}$ | **Strength:** does $A$ really raise the chance of $B$, or are both just popular? ($> 1$ = positive association) |

## 5. Association Rules: Confidence and Lift

$$\text{Rule: } A \longrightarrow B \quad \text{("if } A \text{ is bought, } B \text{ is also bought")}$$

- **Confidence** answers: *"When someone buys $A$, how often do they also buy $B$?"*
- **Lift** answers: *"Does buying $A$ really increase the chance of buying $B$, or is $B$ bought all the time anyway?"*

### Confidence
The **conditional probability** of the consequent $B$ given the antecedent $A$:

$$\text{Confidence}(A \rightarrow B) = P(B \mid A) = \frac{\text{Support}(A \cap B)}{\text{Support}(A)} = \frac{\text{transactions with both } A \text{ and } B}{\text{transactions with } A}$$

- **Range:** 0 to 1.
- **Directional:** $\text{Confidence}(A \rightarrow B)$ is usually **not** equal to $\text{Confidence}(B \rightarrow A)$.

**Weakness:** confidence is misleading when $B$ is very popular. If 90% of shoppers buy bread, almost any rule ending in bread (e.g. Shampoo → Bread) has about 90% confidence, even though shampoo and bread are unrelated. Lift corrects for this.

### Lift
Lift compares how often $A$ and $B$ occur together with how often they would if they were independent:

$$\text{Lift}(A \rightarrow B) = \frac{\text{Confidence}(A \rightarrow B)}{\text{Support}(B)} = \frac{P(A \cap B)}{P(A)\,P(B)}$$

- **Symmetric:** $\text{Lift}(A \rightarrow B) = \text{Lift}(B \rightarrow A)$.
- **Range:** 0 to $\infty$.

| Lift | Relationship | Meaning |
| --- | --- | --- |
| **> 1** | Positive association | Buying $A$ makes $B$ more likely — a useful rule |
| **= 1** | Independence | $A$ and $B$ co-occur only as often as chance predicts |
| **< 1** | Negative association / substitutes | Buying $A$ makes $B$ less likely (e.g. competing brands) |

### Worked example
A store records **100 transactions**: 40 contain **burgers** ($A$), 50 contain **soda** ($B$), 30 contain **both**.

**Step 1 – supports**
- $\text{Support}(A) = 40/100 = 0.40$
- $\text{Support}(B) = 50/100 = 0.50$
- $\text{Support}(A \cap B) = 30/100 = 0.30$

**Step 2 – confidence**
$$\text{Confidence}(\text{Burger} \rightarrow \text{Soda}) = \frac{0.30}{0.40} = 0.75 \;(75\%)$$
Of all customers who bought burgers, **75%** also bought soda.

**Step 3 – lift**
$$\text{Lift}(\text{Burger} \rightarrow \text{Soda}) = \frac{0.75}{0.50} = \mathbf{1.5}$$
A burger buyer is **1.5 times as likely** to buy soda as an average shopper. Lift > 1 → a genuine positive association.

### Summary

| Metric | Formula | Range | Question answered |
| --- | --- | --- | --- |
| **Support** | $P(A \cap B)$ | 0 – 1 | How common is the combination overall? |
| **Confidence** | $\dfrac{P(A \cap B)}{P(A)}$ | 0 – 1 | If $A$ happens, how reliably does $B$ follow? |
| **Lift** | $\dfrac{P(A \cap B)}{P(A)\,P(B)}$ | 0 – ∞ | Is the co-occurrence higher than chance? |

## 6. Classification Metrics

Choose the metric based on the **class balance** and the **cost of each type of error**.

### 6.1 The confusion matrix
Every standard metric is built from the 2×2 confusion matrix (binary classification):

| | Predicted negative (0) | Predicted positive (1) |
| --- | --- | --- |
| **Actual negative (0)** | **TN** — true negative | **FP** — false positive (Type I error) |
| **Actual positive (1)** | **FN** — false negative (Type II error) | **TP** — true positive |

### 6.2 Metrics at a single threshold
These evaluate the predicted labels at one decision threshold (usually 0.5).

**Accuracy**
$$\text{Accuracy} = \frac{TP + TN}{TP + TN + FP + FN}$$
- Share of all predictions that are correct.
- Use only when classes are **balanced**.
- *Pitfall:* if 99% of transactions are legitimate, a model that always says "legitimate" scores 99% accuracy while catching no fraud.

**Precision** (positive predictive value)
$$\text{Precision} = \frac{TP}{TP + FP}$$
- Of everything predicted positive, how much really is positive?
- Optimise when **false positives are costly** — e.g. a spam filter shouldn't send an important email to spam.

**Recall / sensitivity** (true positive rate)
$$\text{Recall} = \frac{TP}{TP + FN}$$
- Of all actual positives, how many did the model catch?
- Optimise when **false negatives are dangerous** — e.g. disease, fraud or intrusion detection.

**Specificity** (true negative rate)
$$\text{Specificity} = \frac{TN}{TN + FP}$$
- Of all actual negatives, how many were correctly labelled negative?

**F1-score and $F_\beta$**
$$\text{F1} = 2 \times \frac{\text{Precision} \times \text{Recall}}{\text{Precision} + \text{Recall}} = \frac{2TP}{2TP + FP + FN}$$
- Harmonic mean of precision and recall; low if either one is low.
- A good single number for **imbalanced** data.
- $F_\beta$ weights recall $\beta$ times as much as precision: $F_2$ favours recall, $F_{0.5}$ favours precision.

### 6.3 Threshold-independent metrics (probability-based)
These assess the predicted **probabilities** across all possible thresholds.

**ROC curve & ROC-AUC**
- Plots the **true positive rate** (recall) against the **false positive rate** ($1 - \text{specificity}$) as the threshold varies.
- AUC = 1.0 → perfect separation; AUC = 0.5 → no better than a coin toss.
- Best for balanced or moderately imbalanced data.

**Precision–recall curve & PR-AUC**
- Plots precision against recall as the threshold varies.
- Best for **severely imbalanced** data (fraud, rare failures): unlike ROC-AUC, it isn't inflated by a huge number of true negatives.

**Log-loss (cross-entropy)**
$$\text{Log-loss} = -\frac{1}{N} \sum_{i=1}^{N} \left[ y_i \log(\hat{p}_i) + (1 - y_i) \log(1 - \hat{p}_i) \right]$$
- Measures how good and how well-calibrated the probabilities are; heavily penalises confident wrong predictions (e.g. 99% for a sample that is actually 0).

### 6.4 Multi-class averaging
With 3+ classes, precision, recall and F1 are computed per class and then averaged:

- **Macro average:** unweighted mean over classes — every class counts equally, so poor minority-class performance shows up.
- **Weighted average:** mean weighted by each class's number of true instances (support).
- **Micro average:** pool all TP, FP and FN first, then compute the metric. In single-label multi-class problems, micro-F1 equals accuracy.

### 6.5 Choosing a metric

```text
Is the dataset balanced?
│
├── YES ──► Accuracy, F1-score, ROC-AUC
│
└── NO (imbalanced)
     │
     ├── False negatives worse? (missed fraud / defect) ──► Recall, F2-score, PR-AUC
     ├── False positives worse? (spam blocking, alarms) ──► Precision, F0.5-score
     └── Need well-calibrated probabilities?            ──► Log-loss, Brier score
```
