# 🤖 Introducing scikit-learn

## 1. What Is scikit-learn?
scikit-learn is a general-purpose Python library for machine learning. It supports supervised and unsupervised learning and offers **one consistent interface** for transforming data, training models, making predictions and validating results.

---

## 2. Supervised Learning Vocabulary

Supervised learning estimates a **target** from a set of **predictors**. Statistics and machine learning use different names for the same ideas:

| Statistics | Machine learning | Meaning |
| --- | --- | --- |
| Independent variable | Predictor / feature | Input used to make the estimate |
| Dependent variable | Target / label | Value to be predicted |

A **label** is the known target value of one example. Any column can act as predictor or target — it depends on the question being asked.

---

## 3. Feature Matrix and Target Array

| | Symbol | Shape | Usual type |
| --- | --- | --- | --- |
| **Feature matrix** | `X` | `(n_samples, n_features)` — always **2-D** | pandas `DataFrame` or 2-D NumPy array |
| **Target array** | `y` | `(n_samples,)` — 2-D only for multi-target problems | pandas `Series` or 1-D NumPy array |

Even with a single feature, `X` must be 2-D:

## 4. Basic Analytics Pipeline

| Step | Scope | What happens |
| --- | --- | --- |
| 1. Data preparation | Project-specific | Load labelled data; build `X` (and `y` for supervised methods) |
| 2. Model & hyperparameter selection | Model-specific | Choose a method (regression, decision tree, clustering, …) and its settings (tree depth, number of clusters, …) |
| 3. Training | General | `model.fit(X, y)` |
| 4. Model display | Model-specific | Inspect coefficients, plot the tree or clusters |
| 5. Application | General | `model.predict(X_new)` or `model.transform(X_new)` |

## 5. Feature Scaling

Scaling puts features with different units and ranges on a comparable scale, so distance- and gradient-based algorithms (K-Means, SVM, neural networks, regularised regression) don't let large-valued features dominate.

| Scaler | Result | Note |
| --- | --- | --- |
| `StandardScaler` | mean = 0, standard deviation = 1 (*standardisation*) | Good default |
| `MinMaxScaler` | values in $[0, 1]$ (*normalisation*) | Sensitive to outliers |

### Scaler methods
- **`fit()`** — learns the scaling parameters (mean/std or min/max).
- **`transform()`** — applies them.
- **`fit_transform()`** — both in one step (training data only).
- **`inverse_transform()`** — converts scaled values back to the original units.

The output is a NumPy array of shape `(n_samples, n_features)`. To keep column names, wrap it in a DataFrame or ask scikit-learn for pandas output:

```python
from sklearn.preprocessing import StandardScaler
import pandas as pd

scaler = StandardScaler()
X_scaled = pd.DataFrame(scaler.fit_transform(X), columns=X.columns, index=X.index)

# or, scikit-learn ≥ 1.2:
scaler = StandardScaler().set_output(transform="pandas")
X_scaled = scaler.fit_transform(X)
```
