# 🔍 Clustering (K-Means)

## 1. What Is Clustering?

Clustering is an **unsupervised** technique that groups similar data points by their features. It reveals natural groupings in data without using any labels.

---

## 2. How K-Means Works

1. Choose the number of clusters $k$ (`n_clusters`).
2. Place $k$ initial centroids (by default with the smart `k-means++` method).
3. **Assign** every point to its nearest centroid.
4. **Update** each centroid to the mean of its assigned points.
5. Repeat 3–4 until the centroids stop moving (or `max_iter` is reached).

K-Means minimises the **inertia** — the sum of squared distances between points and their centroid (within-cluster variance).

⚠️ K-Means uses distances, so **scale the features first** (e.g. `StandardScaler`). It works best for compact, roughly round clusters of similar size; DBSCAN or hierarchical clustering suit other shapes.

---

## 3. Key Hyperparameters (`sklearn.cluster.KMeans`)

| Hyperparameter | Default | Description |
| --- | --- | --- |
| `n_clusters` | `8` | Number of clusters $k$ |
| `init` | `'k-means++'` | Centroid initialisation (`'k-means++'` or `'random'`) |
| `n_init` | `'auto'` | Number of runs with different starting centroids; the best (lowest inertia) is kept. `'auto'` = 1 run with `k-means++`, 10 with `random` |
| `max_iter` | `300` | Maximum iterations per run |
| `tol` | `1e-4` | Convergence tolerance |
| `algorithm` | `'lloyd'` | `'lloyd'` (classic) or `'elkan'` (can be faster on well-separated clusters) |
| `random_state` | `None` | Seed for reproducible results |

---

## 4. Choosing $k$

- **Elbow method:** plot inertia for $k = 1, 2, \dots$; pick the $k$ where the curve bends and further clusters give little improvement.
- **Silhouette score** (`sklearn.metrics.silhouette_score`, range −1 to 1): how well each point fits its own cluster compared with the nearest other cluster — higher is better.

---

## 5. Workflow

1. Generate or load data — `make_blobs` creates synthetic data with known clusters, handy for practice.
2. Create `KMeans` with the chosen hyperparameters.
3. Fit the model.
4. Read the **centroids** (`cluster_centers_`) and the **labels** (`labels_`).
5. Visualise: a scatter plot coloured by cluster, with the centroids overlaid in red.
6. Assign new points with `model.predict(X_new)`.

```python
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.cluster import KMeans
from sklearn.datasets import make_blobs

X, _ = make_blobs(n_samples=300, centers=4, cluster_std=0.8, random_state=42)

model = KMeans(n_clusters=4, random_state=42).fit(X)
labels = model.labels_
centroids = model.cluster_centers_

sns.scatterplot(x=X[:, 0], y=X[:, 1], hue=labels, palette="tab10")
plt.scatter(centroids[:, 0], centroids[:, 1], c="red", marker="X", s=200)
plt.show()

print(model.predict([[0, 2], [-5, 8]]))  # cluster of two new points
```

---

## ✅ Summary
- K-Means is a fast, simple method for grouping data into $k$ clusters.
- Scale features first, and choose $k$ with the elbow method or silhouette score.
- Typical uses: exploratory analysis, customer segmentation, pattern discovery.
